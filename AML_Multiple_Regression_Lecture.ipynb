```python
import os
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
import joblib

print("🎓 CLOUD REGRESSION MODEL RUNNING SUCCESSFULLY")
print("===============================================")

# 1. Dataset parameters using your exact data
# Written explicitly to prevent the system from stripping the numbers
hours_list = [1, 2, 3, 4, 5,6]
grade_list = [50, 60, 70, 80, 90,100]

data = pd.DataFrame({
    'Hours': hours_list,
    'Grade': grade_list
})

X = data[['Hours']]
y = data['Grade']

# 2. Train the Linear Regression Model
model = LinearRegression()
model.fit(X, y)

print(f"✅ Model Trained: Slope is {model.coef_[0]:.2f}")
print(f"🏠 Baseline Score (Y-Intercept): {model.intercept_:.2f}")

# 3. Save the Model File inside the Cloud Workspace
model_filename = "student_linear_model.pkl"
joblib.dump(model, model_filename)

print("🚀 SUCCESS: Saved to cloud!")
```

```python
#@title 🔮 Live Student Grade Predictor
import joblib
import numpy as np

# 1. Load the model we just saved
model = joblib.load("student_linear_model.pkl")

# 2. Interactive Slider for Students
#@markdown Move the slider below to change the study hours and watch the model predict the grade!
study_hours = 0 #@param {type:"slider", min:0, max:10, step:0.5}

# 3. Calculate prediction
predicted_grade = model.predict([[study_hours]])

print("==================================================")
print(f"📊 LIVE DEMONSTRATION FOR STUDENTS")
print("==================================================")
print(f"🎒 Input Study Time : {study_hours} Hours")
print(f"📈 Formula used     : Grade = (10 * {study_hours}) + 40")
print("--------------------------------------------------")
print(f"🔮 PREDICTED GRADE  : {predicted_grade[0]:.1f}%")
print("==================================================")
```

```python
#@title 🛡️ AML Multiple Linear Regression Dashboard
import pandas as pd
from sklearn.linear_model import LinearRegression
import warnings
warnings.filterwarnings("ignore")

print("==================================================")
print("🛡️ AML MULTIPLE REGRESSION ENGINE BOOTING...")
print("==================================================")

# 1. Historical Dataset explicitly written to prevent rendering issues
# X1 = Amount in thousands ($1k, $2k, $5k, $8k, $9k)
amount_data = [1, 2, 5, 8, 9]

# X2 = Number of transfers in a single week
frequency_data = [1, 2, 1, 5, 6]

# Y = Historical Risk Score out of 100 assigned by compliance
risk_score_data = [15, 25, 45, 85, 95]

# Convert lists into a unified DataFrame
data = pd.DataFrame({
    'Amount_K': amount_data,
    'Transfers_Week': frequency_data,
    'Risk_Score': risk_score_data
})

# Separate the two Independent Features (X) from the Target (y)
X = data[['Amount_K', 'Transfers_Week']]
y = data['Risk_Score']

# 2. Train the Multiple Linear Regression Model
model = LinearRegression()
model.fit(X, y)

# 3. Interactive Input Sliders for Classroom Demo
#@markdown Adjust both indicators to see how the model calculates risk in real-time:
Transfer_Amount_Thousands = 8.5 #@param {type:"slider", min:1, max:10, step:0.5}
Transfers_Per_Week = 2 #@param {type:"slider", min:1, max:10, step:1}

# 4. Calculate Live Prediction
new_transaction = [[Transfer_Amount_Thousands, Transfers_Per_Week]]
predicted_risk = model.predict(new_transaction)

# Clip the results mathematically between 0 and 100 for proper reporting
predicted_risk = max(0, min(100, predicted_risk[0]))

# ---- SAVE WITH YOUR NEW DETAILED LECTURE FILE NAME ----
import joblib
joblib.dump(model, "AML_Multiple_Regression_Lecture.pkl")
print("🚀 SUCCESS: Saved as AML_Multiple_Regression_Lecture.pkl")
# -----------------------------------------------------

# 5. Display the Parameter breakdown
print("==================================================")
print("📊 MULTIPLE LINEAR REGRESSION PARAMETERS")
print("==================================================")
print(f"🏠 Baseline Intercept (b)   : {model.intercept_:.2f}")
print(f"📈 Slope for Amount (m1)     : {model.coef_[0]:.2f}")
print(f"📈 Slope for Frequency (m2)  : {model.coef_[1]:.2f}")
print("--------------------------------------------------")
print(f"🎒 Inputs: ${Transfer_Amount_Thousands}k Amount | {Transfers_Per_Week} Transfers/Week")
print(f"🔮 PREDICTED AML RISK SCORE : {predicted_risk:.1f}%")
print("==================================================")
```
